```python
import sys

def clasificar(imc, edad):
    if edad < 45:
        if imc < 22:
            return "Bajo"
        else:
            return "Medio"
    else:
        if imc < 22:
            return "Medio"
        else:
            return "Alto"

def procesar(linea):
    partes = linea.split()
    masa, altura, edad = float(partes[0]), float(partes[1]), float(partes[2])

    if not (0 <= masa <= 150):
        return "ERROR"
    if not (0.1 <= altura <= 2.5):
        return "ERROR"
    if not (0 <= edad <= 110):
        return "ERROR"

    imc = masa / (altura ** 2)
    riesgo = clasificar(imc, edad)
    return f"{imc:.1f} {riesgo}"

def main():
    print("Ingresa masa altura edad separados por espacio.")
    print("Presiona Enter en una línea vacía para terminar.\n")
    while True:
        try:
            linea = input()
        except EOFError:
            break
        linea = linea.strip()
        if not linea:
            break
        print(procesar(linea))

if __name__ == "__main__":
    main()
```
